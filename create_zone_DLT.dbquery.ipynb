```sql
/*
Create delta live table for taxi zones
*/

CREATE TABLE workspace.taxi_trip.zones
USING DELTA
LOCATION 's3://databricks-taxi-trip/taxi-zone-lookup/bronze_taxi_zone/'
AS
SELECT *, current_timestamp() AS created_at, input_file_name() AS source_path, uuid() AS record_id
FROM csv.`s3://databricks-taxi-trip/taxi-zone-lookup/`
WITH (header=True, inferSchema=True)
```
