
# Analytics Tables for Business Metrics

```python
%sql

SELECT pickup_zone, pickup_latitude, pickup_longitude, SUM(total_amount) AS total_revenue
FROM workspace.default.taxi_trips_silver
GROUP BY pickup_zone, pickup_latitude, pickup_longitude
LIMIT 10;
```

## Zones Revenue

```python
%sql

CREATE OR REPLACE TABLE workspace.default.taxi_zone_revenue_gold AS
SELECT pickup_zone, SUM(total_amount) AS total_revenue, COUNT(*) AS total_trips
FROM workspace.default.taxi_trips_silver
GROUP BY pickup_zone
ORDER BY total_revenue DESC
```

## Peak Hours

```python
%sql
CREATE OR REPLACE TABLE workspace.default.taxi_peak_hours_gold AS
SELECT pickup_hour, COUNT(*) AS total_trips, SUM(total_amount) AS total_revenue
FROM workspace.default.taxi_trips_silver
GROUP BY pickup_hour
ORDER BY pickup_hour
```

```python
%sql
SELECT * FROM workspace.default.taxi_peak_hours_gold
```

## Pickup and Dropoff Time

```python
%sql
CREATE OR REPLACE TABLE workspace.default.taxi_pickup_dropoff_time_gold AS
SELECT pickup_hour, dropoff_hour, COUNT(*) AS total_trips
FROM workspace.default.taxi_trips_silver
GROUP BY pickup_hour, dropoff_hour
```


## Revenue by Time of Day

```python
%sql
CREATE OR REPLACE TABLE workspace.default.taxi_time_of_day_revenue_gold AS
SELECT
CASE
    WHEN pickup_hour BETWEEN 5 AND 11 THEN 'Morning'
    WHEN pickup_hour BETWEEN 12 AND 17 THEN 'Afternoon'
    WHEN pickup_hour BETWEEN 18 AND 22 THEN 'Evening'
    ELSE 'Night'
END AS time_of_day,
SUM(total_amount) AS total_revenue,
COUNT(*) AS total_trips
FROM workspace.default.taxi_trips_silver
GROUP BY time_of_day
```

## Revenue by Each Zone

```python
%sql
CREATE OR REPLACE TABLE workspace.default.taxi_zone_revenue_gold AS
SELECT pickup_zone, SUM(total_amount) AS total_revenue, COUNT(*) AS total_trips
FROM workspace.default.taxi_trips_silver
GROUP BY pickup_zone
ORDER BY total_revenue DESC
```
