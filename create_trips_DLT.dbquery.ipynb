```sql
/*
Create delta live table for the taxi trips
*/

CREATE TABLE workspace.taxi_trip.trips
USING DELTA
LOCATION 's3://databricks-taxi-trip/taxi-trip-data-raw/bronze_taxi_trip/'
AS
SELECT *, current_timestamp() AS created_at, input_file_name() AS source_path, uuid() AS record_id
FROM csv.`s3://databricks-taxi-trip/taxi-trip-data-raw/`
WITH (header=True, inferSchema=True)
```
